```python
#---a---
print('a) Write a function called calculate_total_sales(game) that takes a single game row (a list) as input and returns the sum of NA, EU, and JP sales. Test it by calling it on the first game in the dataset and printing the result.')
print(' ')
def calculate_total_sales(game_list, game): 
    total_NA = 0
    total_EU = 0
    total_JP = 0
    for i in game_list:
        if i[1] == game: 
            total_NA = total_NA + i[6]
            total_EU = total_EU + i[7]
            total_JP = total_JP + i[8]
    return total_NA + total_EU + total_JP
# Testing: 
result = calculate_total_sales(video_game_sales, 'Wii Sports')
print('Test the function with Wii Sports game value: \n Result:', result)

#--b---
print(' ')
print("b) Write a function called filter_by_genre(data, genre) that takes the full dataset and a genre string as inputs, and returns a new list containing only the games that match that genre. Give the genre parameter a default value of 'Platform'. Test it with and without specifying a genre.")
print(' ')
def filter_by_genre(data, genre = 'Platform'):
    games1 = []
    for i in data: 
        if i[4] == genre: 
            games1.append(i[1])
    return games1
result = filter_by_genre(video_game_sales, 'Sports')
print('Testing with Sports genre: \n Result: ', result)

# -- c---
print(' ')
print('c) Write a function called get_summary(game) that takes a game row and returns a formatted string like: "Wii Sports (2006) - Sports - $82.74M". Use this function inside a loop to print a summary of every game in the dataset.')
print(' ')
def get_summary(game_list, game):
    for i in game_list: 
        if i[1] == game: 
            return f"{i[1]} ({i[3]}) - {i[4]} - ${i[9]}M"
result = get_summary(video_game_sales, 'Wii Sports')
print ('Test the function with Wii Sports game value: \n Result:', result)
```
